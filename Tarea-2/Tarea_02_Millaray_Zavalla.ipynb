##problema 1: Clasificar temperatura

```python
# Problema 1: Clasificador de temperatura

# Recorremos todos los números desde -30 hasta 50
for temperatura in range(-30, 51):

    # Si la temperatura es menor que 0
    if temperatura < 0:
        print(temperatura, "Muy frío")

    # Si la temperatura está entre 0 y 24
    elif temperatura < 25:
        print(temperatura, "Templado")

    # Si la temperatura es 25 o superior
    else:
        print(temperatura, "Caluroso")
```

## Problema 2: Números especiales del robot

```python
# Problema 2: Números especiales del robot

# Recorremos los números desde 1 hasta 100
for numero in range(1, 101):

    # Comprobamos primero si el número es divisible por 10
    if numero % 10 == 0:
        print(numero, "Bip Bip")

    # Si no, comprobamos si es divisible por 5
    elif numero % 5 == 0:
        print(numero, "Bip")
```

## Problema 3: Detector de puntaje gamer

```python
# Problema 3: Detector de puntaje gamer

# Solicitamos los dos puntajes al usuario
puntaje1 = int(input("Ingresa primer puntaje: "))
puntaje2 = int(input("Ingresa segundo puntaje: "))

# Sumamos ambos puntajes
suma = puntaje1 + puntaje2

# Clasificamos el resultado según la suma
if suma < 50:
    print("Resultado: Nivel principiante")

elif suma < 80:
    print("Resultado: Nivel intermedio")

else:
    print("Resultado: Nivel pro")
```

## Problema 4: Calculador de perfil musical

```python
# Problema 4: Calculador de perfil musical

# Solicitamos la edad del usuario
edad = int(input("Ingresa tu edad: "))

# Preguntamos por sus gustos
musica = input("¿Te gusta la música? (si/no): ")
baile = input("¿Te gusta bailar? (si/no): ")

# Convertimos las respuestas en valores booleanos
le_gusta_musica = musica == "si"
le_gusta_bailar = baile == "si"

# Evaluamos el perfil del usuario
if edad >= 18 and le_gusta_musica == True and le_gusta_bailar == True:
    print("Eres una estrella de la fiesta")

elif le_gusta_musica == True and le_gusta_bailar == False:
    print("Te gusta la música, pero no tanto bailar")

else:
    print("Tienes tu propio estilo")
```
